```python
import matplotlib.pyplot as plt
import pandas as pd
from pathlib import Path

if "data" not in globals():
    base_dir = Path.cwd()
    data_path = base_dir / "EMPLOYEES_SYSTEM.csv"
    if not data_path.exists():
        data_path = base_dir.parent / "EMPLOYEES_SYSTEM.csv"
    data = pd.read_csv(data_path).drop_duplicates().copy()

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
data["CurrentSalary"].plot(kind="hist", bins=30, ax=axes[0], color="#176b52", edgecolor="white")
axes[0].set_title("Current salary distribution")
axes[0].set_xlabel("Current salary")
data.corr(numeric_only=True)["CurrentSalary"].drop("CurrentSalary").sort_values().plot(
    kind="barh", ax=axes[1], color="#d97754"
)
axes[1].set_title("Feature correlation with salary")
plt.tight_layout()
plt.show()
```
